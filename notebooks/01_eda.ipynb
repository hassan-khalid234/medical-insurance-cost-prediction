# Phase - 1

## Problem Statement
Health insurers price policies based on expected future claims. Charging every customer the same premium is unsustainable — high-risk customers cost more than they pay in, low-risk customers get overcharged and leave for competitors. The company needs a way to estimate expected annual medical charges for a customer before claims happen, using information available at enrollment.

## Target variable
"charges" — continuous, in USD. This makes it a regression problem, not classification.

## Input features

age — numeric       
sex — categorical (male/female)           
bmi — numeric       
children — numeric (count of dependents)        
smoker — categorical (yes/no)       
region — categorical (northeast/northwest/southeast/southwest)

## Real-world use case
Insurers use models like this for risk-based premium pricing: estimate a baseline expected cost per customer, then set premiums with margin on top. It's also used for underwriting (flagging unusually high-risk applicants for manual review) and portfolio risk assessment (estimating aggregate expected payout across a customer base). Note upfront, honestly, in your writeup: a model like this — especially with smoker as a feature — will show large price gaps by health behavior, and in practice insurers are legally restricted in which features they can price on (this is a real ethical/regulatory tension worth one sentence in your framing, it'll make your writeup stand out).
