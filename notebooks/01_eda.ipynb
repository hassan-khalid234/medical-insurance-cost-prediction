# Phase - 1

## Problem Statement
Health insurers price policies based on expected future claims. Charging every customer the same premium is unsustainable — high-risk customers cost more than they pay in, low-risk customers get overcharged and leave for competitors. The company needs a way to estimate expected annual medical charges for a customer before claims happen, using information available at enrollment.

## Target variable
"charges" — continuous, in USD. This makes it a regression problem, not classification.

## Input features

age — numeric       
sex — categorical (male/female)           
bmi — numeric       
children — numeric (count of dependents)        
smoker — categorical (yes/no)       
region — categorical (northeast/northwest/southeast/southwest)

## Real-world use case
Insurers use models like this for risk-based premium pricing: estimate a baseline expected cost per customer, then set premiums with margin on top. It's also used for underwriting (flagging unusually high-risk applicants for manual review) and portfolio risk assessment (estimating aggregate expected payout across a customer base). Note upfront, honestly, in your writeup: a model like this — especially with smoker as a feature — will show large price gaps by health behavior, and in practice insurers are legally restricted in which features they can price on (this is a real ethical/regulatory tension worth one sentence in your framing, it'll make your writeup stand out).

```python
df['sex'] = df['sex'].map({'male': 1, 'female': 0})
df['smoker'] = df['smoker'].map({'yes': 1, 'no': 0})
df_encoded = pd.get_dummies(df, columns=['region'], drop_first=True,dtype=int)
```

```python
corr = df_encoded.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
```

## Observations discovered
- Smoking status is the dominant driver of charges. Correlation with charges is 0.79 — by far the highest of any feature. The boxplot shows almost zero overlap between smoker/non-smoker interquartile ranges: non-smokers median around $9-10k, smokers median around $34k.
- Age has a moderate positive correlation (0.30), visible as a general upward trend in Chart 1 — but notice your scatter plot shows three distinct diagonal bands, not one cloud. That's not random noise; it lines up with smoker status and (looking at BMI) obesity, meaning age alone doesn't explain the bands — it's an interaction between age, smoker, and bmi.
- BMI has a weak correlation overall (0.20) — but that number undersells it, because BMI's effect is conditional. Chart 2 shows BMI barely moves charges for non-smokers (mostly flat, low cluster), but for smokers, once BMI crosses ~30 (the obesity threshold), charges jump sharply — that's a visible non-linear interaction, not something the correlation coefficient alone captures.-
- sex, children, and region are all weak/negligible (correlations between -0.04 and 0.07) — none of them meaningfully explain variation in charges on their own.
- Practical implication for modeling (sets up Phase 5): since BMI's effect depends on smoker status, a plain linear model on raw features will underfit that interaction. Worth flagging now that an bmi * smoker interaction term could improve the model later, even if your first pass is a straightforward linear regression on the encoded features.
